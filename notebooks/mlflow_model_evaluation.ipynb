```python
from langchain_openai import ChatOpenAI
import mlflow
import pandas as pd

from dsview.extraction.content_extraction import get_summary_generator
from dsview.content.content_loader import TextLoader
```

```python
content_loader = TextLoader("../data/test/autogluon.txt", 50_000)
content_loader.load()
```

```python
llm = ChatOpenAI(temperature=0, model_name="gpt-4o-mini")
summary_generator = get_summary_generator(llm)

with mlflow.start_run() as run:

	model_info = mlflow.langchain.log_model(summary_generator, "Summary generator")

	results = mlflow.evaluate(
		model_info.model_uri,
  		pd.DataFrame([{"content": content_loader.content, "golden_answer": "some answer."}]),
		targets="golden_answer",
		model_type="text-summarization",
	)
```
